```python
import random

print("="*30)
print("Welcome to the Card Game!")
print("="*30)
print("")


ranks=["2","3","4","5","6","7","8","9","10","JACK","QUEEN","KING","ACE"]
suits=["Hearts","Diamonds","Clubs","Spades"]


deck=[f"{rank} of {suit}" for rank in ranks for suit in suits]

random.shuffle(deck)

print("Deck created and shuffled !")
print("Total cards:",len(deck))
print("")


while True:
  try:
      print("")
      number_of_players=int(input("Enter the number of players: "))
      print("")
      if number_of_players<=0:
        print("Number of players must be greater than 0")
        continue
      elif 52%number_of_players!=0:
        print("Invalid number of players")
        print("Cards cannot be distributed equally")
        continue
      else:
        break
  except ValueError:
        print("Please enter a valid number")


cards_per_player=len(deck)//number_of_players
hands=[]
for index in range(number_of_players):
  start=index*cards_per_player
  end=start+cards_per_player
  player_hand=deck[start:end]
  hands.append(player_hand)

print("\nCards distributed successfully!")
print("Each player receives:",cards_per_player,"cards")
print("")


scores=[0]*number_of_players


for round_number in range(1,cards_per_player+1):


  print("\n"+"="*30)
  print(f"           ROUND {round_number}")
  print("="*30)

  played_cards=[]

  for index in range(number_of_players):
    card=random.choice(hands[index])
    hands[index].remove(card)
    played_cards.append(card)

  for index in range(number_of_players):
    print(f"Player {index+1} played: " f"{played_cards[index]}")


  while True:
    try:
        winner=int(input("Enter the round winner: "))
        if winner<1 or winner>number_of_players:
          print("Invalid round winner")
          continue
        else:
          scores[winner-1]+=1
          break
    except ValueError:
        print("Please enter a valid number")
        continue


  print("\nCurrent Scores:")
  for index in range(number_of_players):
    print(f"Player {index+1}: {scores[index]}")

print("\n")
print("=" * 30)
print("        FINAL RESULTS")
print("=" * 30)

print("\nFinal Scores:")
for index in range(number_of_players):
  print(f"Player {index+1}: {scores[index]} rounds won")


highest_score=max(scores)
winners=[]
for index in range(number_of_players):
  if scores[index]==highest_score:
    winners.append(index+1)

print("\n"+"="*30)

if len(winners) == 1:
  print(f"Overall Winner: Player {winners[0]}")
  print(f"Winning Score: {highest_score} rounds")
else:
  print("The game ended in a tie!")
  print("Winning Players:",winners)
  print(f"Winning Score: {highest_score} rounds")
print("="*30)

```
